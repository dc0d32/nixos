# Geometry and trigonometry

You will use `sympy.geometry` and trig tools with exact values.

```python
from sympy import *
from mathkit import *
from sympy.geometry import Point, Segment, Triangle, Circle
init_printing()
x = symbols("x")
```

## Points and segments

Distance comes from coordinates.

```python
A = Point(0, 0)
B = Point(4, 0)
C = Point(1, 3)
AB = Segment(A, B)
AB.length
```

## Triangle facts

Area, perimeter, and angles are available directly.

```python
T = Triangle(A, B, C)
T.area, T.perimeter, T.angles
```

## Circle intersections

Intersections can be points or empty lists.

```python
c1 = Circle(A, 3)
c2 = Circle(Point(3, 0), 2)
c1.intersection(c2)
```

## Pythagoras with exact arithmetic

For a right triangle with legs 3 and 4, the hypotenuse is:

```python
sqrt(3**2 + 4**2)
```

## A triangle sketch with `plot`

Here each line is one side equation in the same axes window.

```python
plot(0, x, 4 - x, x=(-1, 5), y=(-1, 5), labels=["y=0", "y=x", "y=4-x"])
```

## Unit circle exact point

`unit_circle` shows sin/cos projections.

```python
unit_circle(pi/6)
```
