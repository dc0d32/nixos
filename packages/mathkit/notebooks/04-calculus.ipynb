# Calculus

You will study change (derivatives) and accumulation (integrals).

```python
from sympy import *
from mathkit import *
init_printing()
x = symbols("x")
```

## Limits

A limit asks what value a function approaches.

```python
limit((sin(x))/x, x, 0)
```

## Derivative steps

`diff_steps` explains rule choices.

```python
diff_steps("sin(x^2)*exp(3*x)")
```

## Tangent line picture

A tangent is the best local straight-line approximation.

```python
tangent("x^3 - 2*x", at=1)
```

## Explore tangent position

Move the point and watch the slope change.

```python
explore_tangent("x^3 - 2*x", (-3, 3))
```

## Indefinite integral steps

This gives an antiderivative and checks it.

```python
integrate_steps("2*x*cos(x^2)")
```

## Definite integral steps

Bounds produce one exact number.

```python
integrate_steps(x**2, x, bounds=(0, 3))
```

## Riemann sum picture

Rectangles or trapezoids approximate area.

```python
riemann("x^2", 0, 3, n=8, kind="mid")
```

## Explore Riemann sums

Change the number of slices and method.

```python
explore_riemann("sin(x)", 0, pi)
```

## Series (light intro)

A series expansion is a polynomial approximation near a point.

```python
series(exp(x), x, 0, 6)
```
