# Algebra

You will build expressions, transform them, and solve equations.

```python
from sympy import *
from mathkit import *
init_printing()
x, y = symbols("x y")
```

## Symbols and expressions

Symbols stand for unknown values.

```python
expr = (x + 2)**2 - (x - 1)*(x + 4)
expr
```

## Simplify, expand, factor

These are different views of the same maths object.

```python
simplify(expr), expand(expr), factor(expand(expr))
```

## Step-by-step expand and factor

Use the teaching helpers when you want the working.

```python
expand_steps("(x+2)*(x+3)")
factor_steps("x^2 - 9")
```

## Solve a linear equation

`solve_steps` explains balancing moves.

```python
solve_steps("2x + 3 = 11")
```

## Solve a quadratic equation

You get roots and a check step.

```python
solve_steps(Eq(x**2 - 5*x + 6, 0), x)
```

## Plot two functions

Intersections are where outputs match.

```python
f = x**2 - 4
g = x
xs = solve(Eq(f, g), x)
pts = [(float(v), float(f.subs(x, v))) for v in xs]
plot(f, g, x=(-4, 4), labels=["x^2-4", "x"], points=pts), pts
```

## Systems of equations

Use `solve` or `linsolve` for multiple unknowns.

```python
solve((Eq(2*x + y, 7), Eq(x - y, 2)), (x, y)), linsolve([Eq(2*x + y, 7), Eq(x - y, 2)], (x, y))
```
