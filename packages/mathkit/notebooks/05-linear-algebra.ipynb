# Linear algebra

Matrices can represent systems and geometric transformations.

```python
from sympy import *
from mathkit import *
import numpy as np
import matplotlib.pyplot as plt
init_printing()
```

## Build a matrix

Start with a simple 2x2 matrix.

```python
A = Matrix([[2, 1], [1, 3]])
A
```

## Determinant, inverse, rank

These describe key matrix properties.

```python
A.det(), A.inv(), A.rank()
```

## Eigenvalues and eigenvectors

Eigenvectors keep their direction after transformation.

```python
A.eigenvals(), A.eigenvects()
```

## Solve Ax = b (worked 2x2)

Use matrix solve instead of manual substitution.

```python
b = Matrix([7, 8])
sol = A.LUsolve(b)
sol
```

## Check the system solution

Multiply back to confirm.

```python
A * sol, b
```

## Geometric picture: transform the unit square

Columns of `M` tell where basis vectors go.

```python
M = Matrix([[2, 1], [0.5, 1.5]])
square = np.array([[0, 0], [1, 0], [1, 1], [0, 1], [0, 0]], dtype=float)
transformed = np.array([(M @ Matrix(p)).tolist() for p in square], dtype=float).reshape(-1, 2)
fig, ax = plt.subplots(figsize=(6, 6), dpi=120)
ax.plot(square[:, 0], square[:, 1], label="unit square")
ax.plot(transformed[:, 0], transformed[:, 1], label="transformed square")
ax.axhline(0, color="#444", linewidth=1)
ax.axvline(0, color="#444", linewidth=1)
ax.set_aspect("equal", adjustable="box")
ax.grid(True, alpha=0.3)
ax.legend()
fig
```
