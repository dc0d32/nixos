# Numbers and fractions

You will work with exact arithmetic and visual fraction tools.

```python
from sympy import *
from mathkit import *
init_printing()
```

## Exact fractions

Use `Rational(a, b)` instead of decimal approximations.

```python
Rational(2, 3) + Rational(5, 12)
```

## Fraction bar

This shades the part you have.

```python
fraction_bar(3, 5)
```

## Compare several fractions

Bars make ordering easier to see.

```python
compare_fractions((1, 2), (2, 3), (3, 4))
```

## Number line

You can mark points and intervals.

```python
numberline(-2, Rational(1, 2), (1, 3))
```

## Show fraction working

`frac_steps` explains common-denominator methods.

```python
frac_steps("7/12 - 1/4")
```

## Prime factors and gcd/lcm

These tools show the method, not only the result.

```python
prime_factors(360)
gcd_lcm_steps(84, 30)
```

## Long division

Digit-by-digit working is shown.

```python
long_division(7835, 12)
```

## Percent and decimal conversions

Percent means “out of 100”.

```python
percent = 35
fraction = Rational(percent, 100)
decimal = fraction.evalf()
fraction, decimal
```
