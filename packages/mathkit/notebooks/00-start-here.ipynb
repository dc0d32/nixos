The `plot` helper returns a figure.

```python
plot("sin(x)", x=(-pi, pi), labels=["sin(x)"])
```

## A first check

`check(mine, answer)` tells you if two answers are mathematically the same.

```python
check("2x + 3", "3 + 2*x")
```

## mathkit map

| Function | What it is for |
|---|---|
| `check`, `same` | compare your answer with an official one |
| `diff_steps`, `integrate_steps`, `solve_steps` | show calculus/algebra steps |
| `factor_steps`, `expand_steps`, `frac_steps` | show algebra and fraction steps |
| `prime_factors`, `gcd_lcm_steps`, `long_division` | number methods with working |
| `plot`, `tangent`, `riemann`, `unit_circle` | visual maths pictures |
| `fraction_bar`, `compare_fractions`, `numberline` | fraction and number-line visuals |
| `explore_tangent`, `explore_riemann`, `explore_unit_circle`, `explore` | slider-based exploration (or safe fallback) |
