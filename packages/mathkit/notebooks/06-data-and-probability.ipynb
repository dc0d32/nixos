## A small dataset

Use plain Python and NumPy together.

```python
data = [3, 4, 4, 5, 7, 8, 8, 8, 10]
mean_py = sum(data) / len(data)
mean_np = np.mean(data)
median_np = np.median(data)
mode_value = max(set(data), key=data.count)
std_np = np.std(data, ddof=0)
mean_py, mean_np, median_np, mode_value, std_np
```

## Histogram

A histogram shows distribution shape.

```python
rng = np.random.default_rng(20261117)
samples = rng.normal(loc=0.0, scale=1.0, size=400)
fig, ax = plt.subplots(figsize=(7, 4), dpi=120)
ax.hist(samples, bins=20, edgecolor="white")
ax.set_title("Histogram of seeded normal samples")
ax.grid(True, alpha=0.25)
fig
```

## Scatter plot

Scatter plots show relationships between two variables.

```python
x_vals = np.linspace(-3, 3, 40)
y_vals = 0.8 * x_vals + rng.normal(scale=0.8, size=x_vals.size)
fig, ax = plt.subplots(figsize=(6, 4), dpi=120)
ax.scatter(x_vals, y_vals, alpha=0.75)
ax.set_title("Scatter: noisy linear pattern")
ax.grid(True, alpha=0.25)
fig
```

## Two-dice probability with `itertools`

Count outcomes exactly.

```python
outcomes = list(product(range(1, 7), repeat=2))
sum_is_7 = [o for o in outcomes if o[0] + o[1] == 7]
Rational(len(sum_is_7), len(outcomes))
```

## Simulation with a fixed seed

Estimate the chance of rolling at least one 6 in four throws.

```python
rng2 = np.random.default_rng(314159)
trials = 20000
rolls = rng2.integers(1, 7, size=(trials, 4))
success = np.any(rolls == 6, axis=1)
success.mean()
```
